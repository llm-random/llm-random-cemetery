```python
import torch

# Generate random float32
x = torch.tensor(3.75938, dtype=torch.float32)
print(f"float32:{x.item():.5f}")

# Convert to float16
x_f16 = x.to(dtype=torch.float16)
print(f"float16:{x_f16.item():.5f}")

# Convert to bfloat16
x_bf16 = x.to(dtype=torch.bfloat16)
print(f"bfloat16:{x_bf16.item():.5f}")
```
