```python
from PIL import Image

def resize_jpg(input_path, output_path, width=500, height=625):
    """
    Reads a JPG file from input_path, resizes it to the specified width and height,
    and saves the resulting image to output_path.
    """
    img = Image.open(input_path)
    resized_img = img.resize((width, height), Image.LANCZOS)
    resized_img.save(output_path)

if __name__ == "__main__":
    # Example usage:
    resize_jpg("/home/janek/Downloads/zdj_paszportowe_grozny0001.jpg", "/home/janek/Downloads/uw_format_zdj.jpg")
```
