```python
import plotly.express as px
import pandas as pd

def plot_losses(methods_losses: dict):
# Token counts (in billions)
tokens = [1, 2, 4, 6]

# Define losses manually for each method (replace with your own values)
losses_PC = [2.85, 2.65, 2.45, 2.30]
losses_PCD = [2.70, 2.50, 2.35, 2.20]
losses_HPR = [2.95, 2.80, 2.70, 2.55]
losses_HPRD = [2.80, 2.65, 2.50, 2.35]

# Combine into a dataframe
data = []

for method, losses in [
    ("PC", losses_PC),
    ("PCD", losses_PCD),
    ("HPR", losses_HPR),
    ("HPRD", losses_HPRD),
]:
    for t, l in zip(tokens, losses):
        data.append({"Method": method, "Tokens (B)": t, "Loss": l})

df = pd.DataFrame(data)

# Plot
fig = px.line(
    df,
    x="Tokens (B)",
    y="Loss",
    color="Method",
    markers=True,
    title="PC vs HPR - 10% compression; considering distillation",
    labels={"Loss": "Cross-Entropy Loss", "Tokens (B)": "Processed Tokens [B]"},
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis=dict(dtick=1),
    yaxis=dict(range=[1.9, 3.1]),
)

fig.show()
```
