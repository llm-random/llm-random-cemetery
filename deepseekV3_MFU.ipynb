# DeepSeek V3 MFU

```python
def get_mfu(params, tokens, flops_used):
    theoretical_flops = 6 * params * tokens
    mfu = theoretical_flops / flops_used
    return mfu
```

```python
active_params = 37 * 1e9
tokens = 14.8 * 1e12
gpu_hours_used = 2.664 * 1e6
h800_tflops = {16: 750, 8: 1500}
```

```python
flops_used = gpu_hours_used * 3600 * h800_tflops[8] * 1e12
mfu = get_mfu(params=active_params, tokens=tokens, flops_used=flops_used)
print(f'MFU: {mfu:.4f}')
```
